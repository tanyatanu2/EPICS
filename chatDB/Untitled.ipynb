```python
import sqlite3
```

```python
conn = sqlite3.connect("chat_history.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS chat_history (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER,
    mode TEXT,
    user_message TEXT,
    ai_response TEXT,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")

conn.commit()
conn.close()

print("Database created successfully!")
```

```python
def save_chat(user_id, mode, user_message, ai_response):
    conn = sqlite3.connect("chat_history.db")
    cursor = conn.cursor()

    cursor.execute("""
    INSERT INTO chat_history
    (user_id, mode, user_message, ai_response)
    VALUES (?, ?, ?, ?)
    """, (user_id, mode, user_message, ai_response))

    conn.commit()
    conn.close()

    print("Chat saved!")
```

```python
def get_chat_history(user_id):
    conn = sqlite3.connect("chat_history.db")
    cursor = conn.cursor()

    cursor.execute("""
    SELECT user_message, ai_response, created_at
    FROM chat_history
    WHERE user_id = ?
    ORDER BY created_at
    """, (user_id,))

    history = cursor.fetchall()

    conn.close()

    return history
```

```python
save_chat(
    12345,
    "chat",
    "I have a headache",
    "Please take rest and stay hydrated."
)
```

```python
history = get_chat_history(12345)

for chat in history:
    print("User:", chat[0])
    print("AI:", chat[1])
    print("Time:", chat[2])
    print("-------------------")
```
